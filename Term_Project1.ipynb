# **Feature Extraction**

```python
# Now we extract the required feature

match_df[['id','city','winner','total_runs']]
```

```python
# We need only these 4 columns, now reassign this into match_df

match_df = match_df[['id','city','winner','total_runs']]
```

```python
delivery = delivery.rename(columns={'match_id':'id'})
```

```python
# Now join the match_df and delivery by match_id

match_df.merge(delivery,on='id')
```

```python
# Store the above data as new dataframe

delivery_df = match_df.merge(delivery,on='id')
```

```python
# We need only 2nd team data

delivery_df[delivery_df['inning']==2]
```

```python
delivery_df = delivery_df[delivery_df['inning']==2]
```

## **We need these columns or features --> batting_team , bowling_team , city , runs_left , balls_left , wickets_left , total_runs , current_run_rate , required_run_rate , result.**

```python
delivery_df
```

```python
delivery_df['total_runs_y'] = pd.to_numeric(delivery_df['total_runs_y'], errors='coerce')
```

```python
delivery_df['total_runs_y'] = delivery_df['total_runs_y'].fillna(0).astype(int)
```

```python
# In the delivery_df we have a feature of total_runs_y that is showing how many runs made on every ball

delivery_df.groupby('id')['total_runs_y'].cumsum()  # this give us the cummalitive sum of total_runs_y
```

```python
# Now store this into the new feature in delivery_df by the name of current_score

delivery_df['current_score'] = delivery_df.groupby('id')['total_runs_y'].cumsum()
```

```python
delivery_df.head()
```

```python
# Now we need to calculate runs_left that will be calculated as ( total_runs_x - current_score )

delivery_df['runs_left'] = delivery_df['total_runs_x'] - delivery_df['current_score']
```

```python
delivery_df.head()
```

```python
# Now we have to calculate balls_left

delivery_df['balls_left'] = 120 - (delivery_df['over']*6 + delivery_df['ball'])
```

```python
delivery_df.head()
```

```python
# Now we need wickets left this will be calculated from player_dismissed feature

delivery_df['player_dismissed'] = delivery_df['player_dismissed'].fillna("0")
```

```python
delivery_df.head()
```

```python
delivery_df['player_dismissed'] = delivery_df['player_dismissed'].apply(lambda x : x if x == "0" else "1")
delivery_df['player_dismissed'] = delivery_df['player_dismissed'].astype('int')
```

```python
wickets = delivery_df.groupby('id')['player_dismissed'].cumsum().values
delivery_df['wickets'] = 10 - wickets
delivery_df.head()
```

```python
# Now we have to calculate current run rate
# formula for calculate crr = runs/overs

(delivery_df['current_score']*6)/(120 - delivery_df['balls_left'])
```

```python
# Lets store this as crr feature

delivery_df['crr'] = (delivery_df['current_score']*6)/(120 - delivery_df['balls_left'])
```

```python
# Now we need required run rate

delivery_df['rrr'] = (delivery_df['runs_left']*6)/delivery_df['balls_left']
```

```python
delivery_df.head()
```

```python
# Now we have to calculate result.
# So if the winner and batting team is same then result is 1 otherwisw 0. for doing this we have created a function

def result(row) :
  return 1 if row['batting_team'] == row['winner'] else 0
```

```python
delivery_df.apply(result, axis=1)
```

```python
delivery_df['result'] = delivery_df.apply(result, axis=1)   # store this data as result feature
```

```python
delivery_df.head()
```

```python
# Now we extract only those features which are required to us

delivery_df[['batting_team','bowling_team','city','runs_left','balls_left','wickets','total_runs_x','crr','rrr','result']]
```

```python
# Store the above data into a new dataframe

final_df = delivery_df[['batting_team','bowling_team','city','runs_left','balls_left','wickets','total_runs_x','crr','rrr','result']]
```

```python
final_df.sample(final_df.shape[0])  # suffle the data to remove biasness
```

```python
final_df = final_df.sample(final_df.shape[0])
```

```python
final_df.sample()
```

```python
final_df.isnull().sum()
```

```python
# we drop all the null value in city

final_df.dropna(inplace=True)
```

```python
# we also remove those rows where balls_left = 0 because it will create problem to us when we train our model

final_df = final_df[final_df['balls_left']!=0]
```

# **Model Building**

```python
X = final_df.iloc[:,:-1]
Y = final_df.iloc[:,-1]

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=1)
```

```python
X_train
```

```python
# In the above data we have 3 string features so we have to convert them into numerical values
# we have done this by using one-hot encoding

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
trf = ColumnTransformer([
    ('trf',OneHotEncoder(sparse_output=False,drop='first'),['batting_team','bowling_team','city'])
]
,remainder='passthrough')
```

# Creating a pipeline

```python
from sklearn.linear_model import LogisticRegression  # we are using LogisticRegression because we need classification with percentage
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
```

```python
pipe = Pipeline(steps=[
    ('step1',trf),
    ('step2',LogisticRegression(solver='liblinear'))
])
```

```python
# pipe = Pipeline(steps=[
#     ('step1',trf),
#     ('step2',RandomForestClassifier())
# ])
```

```python
pipe.fit(X_train,Y_train)
```

```python
y_pred = pipe.predict(X_test)
```

# Calculate accuracy of our model

```python
# from sklearn.metrics import accuracy_score
# accuracy_score(Y_test,y_pred) --> by randomforest
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(Y_test,y_pred)
```

I choose LogisticRegression instead of randomForest because it provides smoother output rather than randomforest provide baised output towards one team.
